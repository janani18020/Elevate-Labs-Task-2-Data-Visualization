```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/sales_data.csv')

df.head()
```

```python
import matplotlib.pyplot as plt

sales_by_category = df.groupby('Category')['Sales'].sum()

sales_by_category.plot(kind='bar')

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
sales_by_region = df.groupby('Region')['Sales'].sum()

sales_by_region.plot(kind='bar')

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['Date'] = pd.to_datetime(df['Date'])

monthly_sales = df.groupby(df['Date'].dt.to_period('M'))['Sales'].sum()

monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
profit_by_category = df.groupby('Category')['Profit'].sum()

profit_by_category.plot(kind='bar')

plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
```

```python
df = pd.read_csv('/content/drive/MyDrive/sales_data.csv')
```

```python
df['Date'] = pd.to_datetime(df['Date'])

print("Date column prepared successfully")
```

```python
from matplotlib.backends.backend_pdf import PdfPages

pdf_path = '/content/drive/MyDrive/Task2_Sales_Visual_Report.pdf'

pdf = PdfPages(pdf_path)

print("PDF created")
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(11, 8))
plt.axis('off')

plt.text(
    0.5, 0.75,
        'SALES PERFORMANCE ANALYSIS',
            ha='center',
                fontsize=26,
                    fontweight='bold'
                    )

plt.text(
    0.5, 0.62,
                            'Data Visualization and Storytelling',
                                ha='center',
                                    fontsize=18
                                    )

plt.text(
                                        0.5, 0.48,
                                            'Sales Dataset: 1,200 Records | 9 Columns',
                                                ha='center',
                                                    fontsize=14
                                                    )

plt.text(
                                                        0.5, 0.35,
                                                            'Analysis of sales, profit, categories, regions and monthly trends',
                                                                ha='center',
                                                                    fontsize=13
                                                                    )

plt.text(
                                                                        0.5, 0.22,
                                                                            'Tools: Python | Pandas | Matplotlib | Google Colab',
                                                                                ha='center',
                                                                                    fontsize=12
                                                                                    )

pdf.savefig(fig)
plt.close(fig)
print("Title page added")
```

```python
sales_by_category = df.groupby('Category')['Sales'].sum()

fig, ax = plt.subplots(figsize=(11, 8))

sales_by_category.plot(
    kind='bar',
        ax=ax
        )

ax.set_title(
            'Total Sales by Category',
                fontsize=20,
                    fontweight='bold'
                    )

ax.set_xlabel('Category')
ax.set_ylabel('Total Sales')

ax.tick_params(axis='x', rotation=0)

plt.tight_layout()

pdf.savefig(fig)
plt.close(fig)
print("Sales by Category page added")
```

```python
sales_by_region = df.groupby('Region')['Sales'].sum()

fig, ax = plt.subplots(figsize=(11, 8))

sales_by_region.plot(
    kind='bar',
        ax=ax
        )

ax.set_title(
            'Total Sales by Region',
                fontsize=20,
                    fontweight='bold'
                    )

ax.set_xlabel('Region')
ax.set_ylabel('Total Sales')

ax.tick_params(axis='x', rotation=45)

plt.tight_layout()

pdf.savefig(fig)
plt.close(fig)
print("Sales by Region page added")
```

```python
monthly_sales = df.groupby(
      df['Date'].dt.to_period('M')
      )['Sales'].sum()

fig, ax = plt.subplots(figsize=(11, 8))

monthly_sales.plot(
          kind='line',
              marker='o',
                  ax=ax
                  )

ax.set_title(
                      'Monthly Sales Trend',
                          fontsize=20,
                              fontweight='bold'
                              )

ax.set_xlabel('Month')
ax.set_ylabel('Total Sales')

ax.tick_params(axis='x', rotation=45)

plt.tight_layout()

pdf.savefig(fig)
plt.close(fig)

print("Monthly Sales Trend page added")
```

```python
profit_by_category = df.groupby('Category')['Profit'].sum()

fig, ax = plt.subplots(figsize=(11, 8))

profit_by_category.plot(
    kind='bar',
        ax=ax
        )
ax.set_title(
            'Total Profit by Category',
                fontsize=20,
                    fontweight='bold'
                    )

ax.set_xlabel('Category')
ax.set_ylabel('Total Profit')

ax.tick_params(axis='x', rotation=0)
plt.tight_layout()

pdf.savefig(fig)
plt.close(fig)

print("Profit by Category page added")
```

```python
fig = plt.figure(figsize=(11, 8))
plt.axis('off')

plt.text(
    0.5, 0.88,
        'KEY BUSINESS INSIGHTS',
            ha='center',
                fontsize=24,
                    fontweight='bold'
                    )

plt.text(
                        0.10, 0.72,
                            '1. Technology generated the highest total sales among all categories.',
                                fontsize=15
                                )

plt.text(
                                    0.10, 0.60,
                                        '2. North America recorded the highest sales compared with other regions.',
                                            fontsize=15
                                            )

plt.text(
                                                0.10, 0.48,
                                                    '3. Monthly sales fluctuated throughout the period, showing peaks and drops.',
                                                        fontsize=15
                                                        )
plt.text(
                                                            0.10, 0.36,
                                                                '4. Technology also generated the highest total profit.',
                                                                    fontsize=15
                                                                    )

plt.text(
                                                                        0.10, 0.24,
                                                                            '5. Sales and profit performance can help identify priority areas for business growth.',
                                                                                fontsize=15
                                                                                )
pdf.savefig(fig)
plt.close(fig)
print("Key Business Insights page added")
```
